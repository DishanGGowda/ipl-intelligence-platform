```python
from collections import Counter

results = Counter()

for file in yaml_files:
    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    outcome = match["info"].get("outcome", {})

    if "winner" in outcome:
        results["winner"] += 1

    if "result" in outcome:
        results[outcome["result"]] += 1

print(results)
```

```python
from collections import Counter
from pathlib import Path
import yaml

wicket_types = Counter()

for file in Path("../data/cricsheet").glob("*.yaml"):
    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    innings = match.get("innings", [])

    for inning in innings:
        inning_data = list(inning.values())[0]

        for delivery in inning_data.get("deliveries", []):
            ball_data = list(delivery.values())[0]

            if "wicket" in ball_data:
                wicket_types[ball_data["wicket"]["kind"]] += 1

print(wicket_types)
```

```python
dls_matches = []

for file in Path("../data/cricsheet").glob("*.yaml"):
    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    outcome = match["info"].get("outcome", {})

    if "method" in outcome:
        dls_matches.append(
            (
                file.name,
                outcome.get("method")
            )
        )

print("DLS / Special Method Matches:", len(dls_matches))
dls_matches[:10] 
```

```python
super_overs = []

for file in Path("../data/cricsheet").glob("*.yaml"):
    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    if len(match["innings"]) > 2:
        super_overs.append(file.name)

print("Super Over Matches:", len(super_overs))
super_overs[:20]
```

```python
max_balls = 0
max_match = None

for file in Path("../data/cricsheet").glob("*.yaml"):

    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    total = 0

    for inning in match["innings"]:
        inning_data = list(inning.values())[0]
        total += len(inning_data.get("deliveries", []))

    if total > max_balls:
        max_balls = total
        max_match = file.name

print("Match:", max_match)
print("Deliveries:", max_balls)
```

```python
file_091 = None
file_092 = None

for file in Path("../data/cricsheet").glob("*.yaml"):

    with open(file, "r", encoding="utf-8") as f:
        match = yaml.safe_load(f)

    version = match["meta"]["data_version"]

    if version == 0.91 and file_091 is None:
        file_091 = match

    if version == 0.92 and file_092 is None:
        file_092 = match

print(file_091.keys())
print(file_092.keys())
```
